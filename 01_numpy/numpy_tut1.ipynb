```python
import numpy as np
import time
```

```python
# pylist to array
arr=np.array([1,2,3,4])
print(arr,type(arr))
```

```python
# exceution performance time
size=10_000_000

py_list= list(range(size)) #[1,2,3...]
start=time.time()
sq_list = [x**2 for x in py_list]
end = time.time()
print(f"time for pylist execution{end-start}")

np_arr=np.array(py_list)
start=time.time()
#vectorization
np_sq_list= np_arr**2
end=time.time()
print(f"time for np list execution{end-start}")
```

```python
#memory 
import sys
print(f"py list size {sys.getsizeof(py_list) *len(py_list)} bytes")
print(f"numpy list size {np_arr.nbytes} bytes")
```

```python
# creating from list
arr=np.array([1,2,3,4])
print(arr,type(arr),arr.dtype,arr.shape)

arr2=np.array([1,2,3,4,"prime"])
print(arr2,type(arr2),arr2.dtype,arr2.shape) #so here it converts everything into string see '  '
arr2Dimensional=np.array([[1,2,3],[4,5,6],[71,82,93]])
print(arr2Dimensional,type(arr2Dimensional),arr2Dimensional.dtype,arr2Dimensional.shape)
```

```python
# create from scratch
arr3=np.zeros((2,3), dtype="int64") #prefix with zeros #typecaste when list not present 
print(arr3,arr3.dtype,arr3.shape)

arr4=np.ones((2,3))#prefix with ones
print(arr4,arr4.dtype,arr4.shape)

arr5=np.full((2,3),100) #prefix with value
print(arr5,arr5.dtype,arr5.shape)

arr36=np.eye(4) #identity matrix
print(arr36,arr36.dtype,arr36.shape)

# arr6=np.arrange(0,100,4)
# print(arr6,arr6.dtype,arr6.shape) some problem in its wroking

arr46=np.linspace(0,100,5) #creates evenly spaced arrays
print(arr46,arr46.dtype,arr46.shape,arr46.size)
```
