```python
import json
```

```python
#load the data
def load_data(filename):
    with open(filename,"r") as f:
        data=json.load(f)

    return data
```

```python
data= load_data("store_data.json")
print(data)
print(type(data))
```

```python
#clean n structure the data
def clean_data(data):
    text_to_num={"one" : 1,"two": 2, "three" :3 , "four":4, "five":5}
    cleaned_data= []
    unique_users= set()
    
    for user in data:
        raw_rating = str(user["rating"]).strip().lower()
        if(raw_rating in text_to_num):   
            raw_rating= text_to_num[raw_rating]
        user["rating"] = raw_rating
        
        #handle missing data
        raw_age= user.get("age")
        if raw_age==None :
            user["age"]=None

        #deduplication
        if(user["name"].strip() in unique_users):
            continue

        unique_users.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data
        
```

```python
data=clean_data(data)
print(data)
print(len(data))
```

```python
#to get meaning ful insights from data
def get_insights(data):
    total_rating= 0
    for user in data:
        total_rating+= float(user["rating"])
        
    print(f"avg rating= {total_rating/len(data)}")

#poor rating
    poor_rating=0
    for user in data:
        if (float(user["rating"]) < 3) :
            poor_rating+=1

    print(f"poor rating percentage is {poor_rating / len(data) *100}")
        
        
        
```

```python
get_insights(data)

```

```python
#recommendation 
def get_recommedation(data):
    recommendations=[]
    for user in data:
        curr_recomm= {}
        # curr_recomm["name"]= user["name"] or
        curr_recomm = {
        "name": user["name"].strip()
        }
        if (float(user["rating"]) >=4 ):
            curr_recomm["brand"]="APPLE"
        else:
            curr_recomm["brand"]="SAMSUNG"
        recommendations.append(curr_recomm)
    return recommendations
```
